Open question: this ouput is a single item list, hence the `sabol2026['stats'][0]` preceding the `['splits']` key. If Sabol had played in multiple levels, would there be more than one item in that list, or would they all appear under the `['split']` heading? Does the answer differ for minors vs. majors?

*Also, does the below code work for players who played multiple positions in a single game? (Probably that's why it's a list, below code only indexes 0)* UPDATE: Yes; the revised code below handles multiple positions. Note the game in which Sabol moved back and forth between P and LF to see an edge case.

```python
print("Test loop through Blake Sabol's 2026 season!")
print("--------------------------------------------")
for index, game in enumerate(sabol2026['stats'][0]['splits']):
    positions_string = ""
    for i in range(len(game['positionsPlayed'])):
        positions_string += " " + game['positionsPlayed'][i]['abbreviation']
    print(f'Game: {index + 1} leagueid: {game['team']['league']['id']} position:{positions_string}')
```

```python
def get_minor_gls(player_id, season):
    return requests.get(f"https://statsapi.mlb.com/api/v1/people/{player_id}/stats?stats=gameLog&gameType=R&leagueListId=milb_all&group=hitting&hydrate=team(league)&language=en&season={season}",).json()
```
