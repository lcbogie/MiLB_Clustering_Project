```python
print(dbroster)
```

```python
roster_split = dbroster.split()
print(roster_split)
```

```python
def player_name_fetcher(split_roster):
    name_list = []
    for i in range(2, len(split_roster), 4):
        full_name = split_roster[i] + " " + split_roster[i+1]
        name_list.append(full_name)
    return name_list
```

```python
print(player_name_fetcher(roster_split))
```

```python
db = player_name_fetcher(roster_split)
```
